# 8. Instructor Effectiveness Score and Classification

The objective of this project was to classify instructors into three effectiveness categories.

Since the dataset did not contain predefined effectiveness labels, a composite Instructor Effectiveness Score was created.

The score was based on three academic performance indicators:

1. Completion Rate
2. Average Score Improvement
3. Average Quiz Score

Each variable was normalised using Min-Max scaling and given equal importance in the final score.

The three normalised values were combined to calculate the Instructor Effectiveness Score on a common scale.

Dropout rate was not included because it was highly correlated with completion rate.

## Classification of Instructors

The effectiveness scores were divided into three categories using quantile-based classification.

The categories were:

- Low
- Medium
- High

The final dataset contained 40 instructors in each category.

This created a balanced classification problem and allowed us to compare model performance across all three classes.

It is important to note that these categories represent relative performance within the dataset. They are not independently established standards of teaching quality.

# 14. Business Interpretation and Key Questions

## 14.1 Which features most influenced instructor effectiveness and why?

The analysis shows that five features were useful in predicting instructor effectiveness. These were feedback response rate, average feedback score, average watch time, forum activity rate and assignment submission rate.

Among these features, **feedback response rate was the most important**. It had the highest average importance score of 0.142. This suggests that the model depended heavily on this feature when making predictions. In simple terms, instructors who responded to student feedback more often were more likely to be placed in a higher effectiveness category.

The second most important feature was **average feedback score**, with an importance score of 0.1048. This shows that student feedback was useful in identifying differences between instructors. Instructors with better feedback scores were generally more likely to belong to the High effectiveness group.

**Average watch time** was the third most important feature, with a score of 0.068. The results showed that students spent more time watching lessons from instructors in the High effectiveness group. This may suggest that students found their lessons more useful or engaging.

**Forum activity rate** also played an important role, with a score of 0.0561. Instructors with higher forum activity generally had better effectiveness categories. This may be because regular interaction helps students stay involved in the learning process.

Finally, **assignment submission rate** had an importance score of 0.0313. It was less important than the other four features, but it still provided some useful information to the model.

The number of courses handled and batches per course had almost no importance in our final analysis. This suggests that these two features did not help the model much in distinguishing between the three effectiveness categories.

Overall, the results suggest that student interaction, feedback and engagement were more useful for predicting instructor effectiveness than the number of courses or batches handled.

These results show which features helped our model make predictions. They do not prove that improving these features will automatically make an instructor more effective.

## 14.2 Which variables could be misleading or confounded?

Although our model uses five useful features, we should not assume that all of them give a complete picture of an instructor's performance. Some variables can be affected by factors other than the instructor's actual teaching ability.

The first example is **average feedback score**. A high feedback score may suggest that students are happy with an instructor. However, student feedback can also depend on personal preferences. Some students may give higher ratings simply because a course is easy. Others may give lower ratings because they find the subject difficult. So feedback scores should not be treated as a perfect measure of teaching quality.

The second example is **feedback response rate**. Our analysis found that this was the most important feature. However, a high response rate does not always mean that an instructor is more effective. It only tells us that the instructor responds to feedback more often. We do not know whether the responses actually solve student problems or improve their learning experience.

**Average watch time** can also be misleading. Students spending more time watching lessons does not always mean they are learning more. Some students may need extra time to understand a topic. Others may leave videos running without watching them properly. The length and difficulty of the lessons can also affect watch time.

Another variable is **assignment submission rate**. A high submission rate may look positive, but it does not tell us whether students have understood the subject. Students may submit assignments simply to meet deadlines. Their submission behaviour can also depend on the difficulty of the assignments and the time available to complete them.

Similarly, **forum activity rate** may depend on the size of a class. An instructor handling a larger group may naturally have more discussions and interactions. This does not necessarily mean that the instructor is better than someone handling a smaller group.

There is also an important issue with some variables that we removed before training the model. Completion rate and dropout rate were almost perfectly negatively correlated in our dataset, with a correlation of approximately -0.994. This means they carried almost the same information. Keeping both could have given too much importance to the same student outcome. We therefore excluded dropout rate from the final set of predictors.

Overall, these variables are useful, but they should be interpreted carefully. They may reflect student behaviour, course difficulty, class size and other factors along with instructor performance. More information would be needed to separate these effects properly.

Our model predicts effectiveness categories using patterns in the available data. It cannot always identify the reason behind those patterns.

For example, an instructor may have low forum activity because the course is designed around independent learning. That does not automatically make the instructor ineffective.

We should therefore avoid judging instructors using only these numbers.

## 14.3 How could this model fail in real-world usage?

Our model performed reasonably well during testing, but there are several situations where it could fail if used in a real learning environment.

The first problem is the **small size of our dataset**. We trained the model using data from only 96 instructors and tested it on 24 instructors. This is a very small sample. The model may perform differently when it is used with hundreds or thousands of instructors. It may also struggle when it comes across instructors with patterns that were not present in our original dataset.

The second problem is that **student behaviour can change over time**. For example, students may become less active in discussion forums or start watching fewer recorded lessons. These changes may happen because of changes in the course or learning platform. If the model continues using old patterns, its predictions may become less accurate.

Another problem is the possibility of **incorrect predictions**. Our final model achieved 58.33% accuracy on the test data. This means it correctly classified 14 out of 24 instructors. It also made mistakes across all three effectiveness categories. For example, some Medium effectiveness instructors were classified as High or Low. Such mistakes could create an incorrect impression of an instructor's performance.

The model may also fail when it is used for courses with very different teaching methods. For example, a course that requires independent study may have less forum activity. A practical course may have different student engagement patterns compared to a theory-based course. The model may not understand these differences because it was trained on the patterns available in our dataset.

Another limitation is that our effectiveness categories were created using the available data. We divided instructors into Low, Medium and High groups based on their relative scores. These categories do not represent fixed standards of teaching quality. An instructor classified as High in one dataset may not necessarily meet the same standard in another dataset.

Finally, the model may give unfair results if it is used without human review. An instructor should not be judged only on watch time, feedback scores or assignment submission rates. There may be valid reasons behind these numbers that the model cannot understand.

Overall, our model can be useful for identifying patterns and highlighting instructors who may need further attention. However, it should not be treated as a final decision-making system. Its predictions should always be checked with additional information and human judgement.

The biggest concern is not just that our model makes mistakes. It is that people may trust those predictions too much.

For example, classifying an instructor as Low does not automatically mean that the instructor is a poor teacher. It only means that the model found their data similar to the patterns associated with the Low category in our dataset.

That distinction is very important.

## 6. Statistical Outlier Detection

The Interquartile Range (IQR) method is used to identify
potential statistical outliers in numerical variables.

Outlier detection is performed for investigation,
not automatic deletion.

```python
# ==========================================
# 6.1 IQR OUTLIER DETECTION
# ==========================================

numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

# Remove diagnostic columns created earlier
numeric_columns = [
    col for col in numeric_columns
    if col not in [
        'completion_dropout_sum',
        'deviation_from_one'
    ]
]

outlier_results = []

for col in numeric_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    outlier_results.append({
        'Feature': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': len(outliers),
        'Outlier Percentage':
            len(outliers) / len(df) * 100
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary.round(3))
```

```python

# STEP 20 — OUTLIER DETECTION USING IQR

outlier_features = [
    'completion_rate',
    'dropout_rate',
    'avg_quiz_score',
    'avg_score_improvement',
    'engagement_index',
    'academic_performance_index'
]

outlier_summary = []

for col in outlier_features:

    Q1 = instructor_performance[col].quantile(0.25)
    Q3 = instructor_performance[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = instructor_performance[
        (instructor_performance[col] < lower_bound) |
        (instructor_performance[col] > upper_bound)
    ]

    outlier_summary.append({
        'Feature': col,
        'Lower Bound': round(lower_bound, 3),
        'Upper Bound': round(upper_bound, 3),
        'Outlier Count': len(outliers)
    })

outlier_results = pd.DataFrame(outlier_summary)

display(outlier_results)
```

## 11. Instructor Effectiveness Score

An outcome-based scoring framework is developed using
three learner outcome indicators:

1. Completion rate
2. Average score improvement
3. Average quiz score

Min-Max normalization is applied to bring all three
indicators onto a common 0-100 scale.

The normalized indicators receive equal weights.

Dropout rate is excluded from the composite score
because of its extremely high negative correlation
with completion rate (-0.994).

Engagement and feedback indicators are reserved
as independent predictors for the classification model.

```python

# ==========================================
# STEP 23.3 — EFFECTIVENESS SCORE
# ==========================================

import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# Select outcome indicators
scoring_features = [
    'completion_rate',
    'avg_score_improvement',
    'avg_quiz_score'
]

# Normalize outcome indicators to 0–100
scaler_outcome = MinMaxScaler(feature_range=(0, 100))

normalized_outcomes = scaler_outcome.fit_transform(
    instructor_performance[scoring_features]
)

# Store normalized components
normalized_columns = [
    'completion_normalized',
    'improvement_normalized',
    'quiz_normalized'
]

instructor_performance[normalized_columns] = (
    normalized_outcomes
)

# Calculate equally weighted effectiveness score
instructor_performance['effectiveness_score'] = (
    instructor_performance[normalized_columns]
    .mean(axis=1)
)

# Display results
print("INSTRUCTOR EFFECTIVENESS SCORES:")

display(
    instructor_performance[
        ['instructor_id'] +
        normalized_columns +
        ['effectiveness_score']
    ].head(10).round(2)
)

print("\nSCORE DISTRIBUTION:")

display(
    instructor_performance['effectiveness_score']
    .describe()
    .round(2)
)
```

## 12. Instructor Effectiveness Classification

Instructors are classified into three effectiveness
categories using quantile-based segmentation.

Low: Bottom 33.33% of scores.
Medium: Middle 33.33% of scores.
High: Top 33.33% of scores.

This approach creates approximately balanced
classification categories.

```python

# STEP 24 — CREATE CLASSIFICATION TARGET

instructor_performance['effectiveness_category'] = pd.qcut(
    instructor_performance['effectiveness_score']
        .rank(method='first'),
    q=3,
    labels=['Low', 'Medium', 'High']
)

# Display classification results
print("EFFECTIVENESS CATEGORY DISTRIBUTION:")

display(
    instructor_performance[
        'effectiveness_category'
    ].value_counts().sort_index()
)

# Check percentages
print("\nCATEGORY DISTRIBUTION (%):")

display(
    (instructor_performance[
        'effectiveness_category'
    ].value_counts(normalize=True) * 100)
    .sort_index()
    .round(2)
)
```

## 13. Feature Selection and Data Leakage Prevention

Predictor variables are selected to evaluate instructor
effectiveness using engagement, feedback and workload.

Variables directly used to construct the effectiveness
score are excluded from model predictors.

Derived indices containing overlapping information
are also excluded.

```python

# STEP 25 — SELECT MODEL FEATURES

target = 'effectiveness_category'

excluded_features = [
    'instructor_id',
    'completion_rate',
    'avg_score_improvement',
    'avg_quiz_score',
    'dropout_rate',
    'completion_normalized',
    'improvement_normalized',
    'quiz_normalized',
    'effectiveness_score',
    'academic_performance_index'
]

X = instructor_performance.drop(
    columns=excluded_features + [target],
    errors='ignore'
)

y = instructor_performance[target]

print("PREDICTOR FEATURES:")
print(X.columns.tolist())

print("\nNUMBER OF PREDICTORS:", X.shape[1])
print("NUMBER OF OBSERVATIONS:", X.shape[0])

print("\nTARGET DISTRIBUTION:")
print(y.value_counts())
```

## 14. Train-Test Split

The dataset is divided into training and testing subsets.

80% of observations are allocated for training
and 20% for testing.

Stratified sampling preserves the distribution
of effectiveness categories in both subsets.

```python

# STEP 27 — TRAIN TEST SPLIT

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("TRAINING DATA SHAPE:", X_train.shape)
print("TESTING DATA SHAPE:", X_test.shape)

print("\nTRAINING CLASS DISTRIBUTION:")
print(y_train.value_counts())

print("\nTESTING CLASS DISTRIBUTION:")
print(y_test.value_counts())
```

```python

# STEP - CROSS-VALIDATED PERMUTATION FEATURE IMPORTANCE

from sklearn.inspection import permutation_importance
from sklearn.base import clone
from sklearn.model_selection import RepeatedStratifiedKFold
import pandas as pd
import numpy as np

# Store feature importance from every validation round
importance_results = []

# Repeated stratified cross-validation
importance_cv = RepeatedStratifiedKFold(
    n_splits=5,
    n_repeats=10,
    random_state=42
)

# Run 50 validation rounds
for fold_number, (train_idx, val_idx) in enumerate(
    importance_cv.split(X_train, y_train), 1
):

    # Separate training and validation observations
    X_fold_train = X_train.iloc[train_idx]
    X_fold_val = X_train.iloc[val_idx]

    y_fold_train = y_train.iloc[train_idx]
    y_fold_val = y_train.iloc[val_idx]

    # Create a fresh model
    fold_model = clone(stable_svm)

    # Train model
    fold_model.fit(X_fold_train, y_fold_train)

    # Calculate permutation importance
    perm = permutation_importance(
        fold_model,
        X_fold_val,
        y_fold_val,
        n_repeats=10,
        random_state=42 + fold_number,
        scoring='f1_macro'
    )

    # Store mean importance for this fold
    importance_results.append(
        perm.importances_mean
    )

# Convert results into DataFrame
importance_matrix = pd.DataFrame(
    importance_results,
    columns=X_train.columns
)

# Calculate average importance and variability
importance_summary = pd.DataFrame({
    'Feature': X_train.columns,
    'Mean Importance': importance_matrix.mean().values,
    'Std Dev': importance_matrix.std().values,
    'Positive Folds': (importance_matrix > 0).sum().values
})

# Calculate percentage of folds with positive importance
importance_summary['Positive Fold %'] = (
    importance_summary['Positive Folds'] / 50 * 100
).round(2)

# Sort by average importance
importance_summary = importance_summary.sort_values(
    'Mean Importance',
    ascending=False
)

print("CROSS-VALIDATED FEATURE IMPORTANCE")
print("----------------------------------")

display(importance_summary.round(4))
```

```python

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

from sklearn.model_selection import (
    GridSearchCV,
    StratifiedKFold,
    RepeatedStratifiedKFold
)

from sklearn.metrics import accuracy_score, f1_score

# ==========================================
# 1. VALIDATION SETUP
# ==========================================

outer_cv = RepeatedStratifiedKFold(
    n_splits=5,
    n_repeats=3,
    random_state=42
)

inner_cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

# ==========================================
# 2. DEFINE THREE MODEL PIPELINES
# ==========================================

models = {

    'Logistic Regression': {
        'pipeline': Pipeline([
            ('selector', SelectKBest(f_classif)),
            ('scaler', StandardScaler()),
            ('model', LogisticRegression(max_iter=2000))
        ]),
        'params': {
            'selector__k': [5, 7],
            'model__C': [0.1, 1, 10, 100]
        }
    },

    'Random Forest': {
        'pipeline': Pipeline([
            ('selector', SelectKBest(f_classif)),
            ('scaler', 'passthrough'),
            ('model', RandomForestClassifier(random_state=42))
        ]),
        'params': {
            'selector__k': [5, 7],
            'model__n_estimators': [200, 400],
            'model__max_depth': [None, 5, 10],
            'model__min_samples_leaf': [1, 2]
        }
    },

    'SVM': {
        'pipeline': Pipeline([
            ('selector', SelectKBest(f_classif)),
            ('scaler', StandardScaler()),
            ('model', SVC())
        ]),
        'params': {
            'selector__k': [5, 7],
            'model__C': [0.1, 1, 10],
            'model__gamma': ['scale', 0.01, 0.1],
            'model__kernel': ['rbf', 'linear']
        }
    }
}

# ==========================================
# 3. NESTED CROSS-VALIDATION
# ==========================================

nested_results = []

for model_name, config in models.items():

    print(f"\nRUNNING: {model_name}")
    print("-" * 40)

    fold_number = 0

    for train_idx, val_idx in outer_cv.split(
        X_train, y_train
    ):

        fold_number += 1

        X_outer_train = X_train.iloc[train_idx]
        X_outer_val = X_train.iloc[val_idx]

        y_outer_train = y_train.iloc[train_idx]
        y_outer_val = y_train.iloc[val_idx]

        search = GridSearchCV(
            estimator=config['pipeline'],
            param_grid=config['params'],
            cv=inner_cv,
            scoring='f1_macro',
            n_jobs=-1,
            refit=True
        )

        search.fit(X_outer_train, y_outer_train)

        predictions = search.best_estimator_.predict(
            X_outer_val
        )

        accuracy = accuracy_score(
            y_outer_val, predictions
        )

        macro_f1 = f1_score(
            y_outer_val,
            predictions,
            average='macro'
        )

        nested_results.append({
            'Model': model_name,
            'Fold': fold_number,
            'Accuracy': accuracy,
            'Macro F1': macro_f1,
            'Best Parameters': search.best_params_
        })

        print(
            f"Fold {fold_number}: "
            f"Accuracy = {accuracy*100:.2f}% | "
            f"Macro F1 = {macro_f1:.3f}"
        )

# ==========================================
# 4. SUMMARISE RESULTS
# ==========================================

nested_df = pd.DataFrame(nested_results)

summary = nested_df.groupby('Model').agg(
    Mean_Accuracy=('Accuracy', 'mean'),
    Accuracy_Std=('Accuracy', 'std'),
    Mean_Macro_F1=('Macro F1', 'mean'),
    Macro_F1_Std=('Macro F1', 'std')
).sort_values(
    'Mean_Macro_F1',
    ascending=False
)

print("\nNESTED CROSS-VALIDATION RESULTS")
print("=" * 50)

display((summary * 100).round(2))

# Display best parameters from each outer fold
print("\nBEST PARAMETERS BY MODEL AND FOLD")

for model_name in models:
    print(f"\n{model_name}")
    display(
        nested_df[
            nested_df['Model'] == model_name
        ][['Fold', 'Best Parameters']]
    )
```

```python
import time
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

from sklearn.model_selection import (
    RepeatedStratifiedKFold,
    StratifiedKFold,
    GridSearchCV
)

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    balanced_accuracy_score
)

# Identical validation splits for both models
outer_cv = RepeatedStratifiedKFold(
    n_splits=5,
    n_repeats=3,
    random_state=42
)

inner_cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

# Logistic Regression pipeline
lr_pipeline = Pipeline([
    ('selector', SelectKBest(f_classif)),
    ('scaler', StandardScaler()),
    ('model', LogisticRegression(max_iter=2000, random_state=42))
])

lr_grid = {
    'selector__k': [5, 7],
    'model__C': [0.1, 1, 10, 100]
}

# SVM pipeline
svm_pipeline = Pipeline([
    ('selector', SelectKBest(f_classif)),
    ('scaler', StandardScaler()),
    ('model', SVC())
])

svm_grid = {
    'selector__k': [5, 7],
    'model__C': [0.1, 1, 10],
    'model__gamma': ['scale', 0.01, 0.1],
    'model__kernel': ['rbf', 'linear']
}

candidates = {
    'Logistic Regression': (lr_pipeline, lr_grid),
    'SVM': (svm_pipeline, svm_grid)
}

results = []

# Generate the folds once so both models receive identical splits
splits = list(outer_cv.split(X_train, y_train))

for model_name, (pipeline, grid) in candidates.items():

    print(f"\nFIGHTING: {model_name}")
    start_time = time.time()

    for fold_number, (train_idx, val_idx) in enumerate(splits, 1):

        X_fold_train = X_train.iloc[train_idx]
        X_fold_val = X_train.iloc[val_idx]

        y_fold_train = y_train.iloc[train_idx]
        y_fold_val = y_train.iloc[val_idx]

        search = GridSearchCV(
            estimator=clone(pipeline),
            param_grid=grid,
            cv=inner_cv,
            scoring='f1_macro',
            n_jobs=-1,
            refit=True
        )

        search.fit(X_fold_train, y_fold_train)

        predictions = search.predict(X_fold_val)

        results.append({
            'Model': model_name,
            'Fold': fold_number,
            'Accuracy': accuracy_score(y_fold_val, predictions),
            'Macro_F1': f1_score(
                y_fold_val, predictions, average='macro'
            ),
            'Balanced_Accuracy': balanced_accuracy_score(
                y_fold_val, predictions
            ),
            'Best_Params': search.best_params_
        })

        print(
            f"Fold {fold_number}/15 | "
            f"Macro F1: "
            f"{f1_score(y_fold_val, predictions, average='macro'):.3f}"
        )

    print(f"Time taken: {(time.time() - start_time)/60:.2f} minutes")

# Store results
duel_df = pd.DataFrame(results)

# Summary comparison
duel_summary = duel_df.groupby('Model').agg(
    Mean_Accuracy=('Accuracy', 'mean'),
    Accuracy_SD=('Accuracy', 'std'),
    Mean_Macro_F1=('Macro_F1', 'mean'),
    Macro_F1_SD=('Macro_F1', 'std'),
    Mean_Balanced_Accuracy=('Balanced_Accuracy', 'mean')
)

print("\nFINAL HEAD-TO-HEAD RESULTS")
display((duel_summary * 100).round(2))

# Paired fold-by-fold differences
lr_scores = duel_df[
    duel_df['Model'] == 'Logistic Regression'
].sort_values('Fold')['Macro_F1'].to_numpy()

svm_scores = duel_df[
    duel_df['Model'] == 'SVM'
].sort_values('Fold')['Macro_F1'].to_numpy()

differences = svm_scores - lr_scores

print("\nPAIRED COMPARISON (SVM - LOGISTIC REGRESSION)")
print("Mean Macro F1 difference:", round(differences.mean(), 4))
print("SVM wins:", np.sum(differences > 0), "/ 15")
print("Logistic Regression wins:", np.sum(differences < 0), "/ 15")
print("Ties:", np.sum(differences == 0), "/ 15")
```
